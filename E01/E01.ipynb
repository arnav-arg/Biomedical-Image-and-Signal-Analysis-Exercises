## 1.2 Visualize with Python

```python
import matplotlib.pyplot as plt


def impulse_function(n: int) -> float:
    # Our default value is 0
    y = 0.0
    if n == 0:
        y = 1.0
    return y


n = range(-5, 6)
x_1 = []
x_2 = []

for value in n:
    x_1.append(impulse_function(value - 3))
    x_2.append(impulse_function(value - 5))

for title, x in [("x_1", x_1), ("x_2", x_2)]:
    plt.title(f"${title}$")
    plt.stem(n, x)
    plt.xlim(n[0], n[-1] + 1)
    plt.xlabel("$n$")
    plt.ylabel("$x[n]$")

    plt.show()
```
