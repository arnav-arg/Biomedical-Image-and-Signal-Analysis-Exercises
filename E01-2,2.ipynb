```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x0 = 1 #Amplitude in volts
```

```python
t = np.arange(-10, 10, 0.1)
# Equidistant intervals in the range of -10 to 10 in 0.1 steps
n = np.arange(-10, 11)  # Time index
```

```python
# Continuous-time signals
x_5_t = x0 * np.sin(2 * np.pi * t / 10)
x_6_t = x0 * np.cos(t / 3)
```

```python
# Discrete-time signals
x_5_n = x0 * np.sin(2 * np.pi * n / 10)
x_6_n = x0 * np.cos(n / 3)
```

```python
for title, x_t, x_n in [("x_5", x_5_t, x_5_n),
                         ("x_6", x_6_t, x_6_n)]:

    plt.suptitle(f"${title}$")

    # Graph x_t
    plt.subplot(2, 1, 1)
    plt.plot(t, x_t, linewidth=2)
    plt.grid(True)
    plt.xlabel("$t$")
    plt.ylabel("$x(t)$")

    # Graph x_n
    plt.subplot(2, 1, 2)
    plt.stem(n, x_n)
    plt.grid(True)
    plt.xlabel("$n$")
    plt.ylabel("$x[n]$")

    plt.show()
```
